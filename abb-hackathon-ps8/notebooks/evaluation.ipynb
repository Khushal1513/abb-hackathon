# Phase 1 Evaluation — NLU Intent Routing & Entity Extraction
This notebook verifies the `NLUProblemParser` pipeline against the sample problem statement scenario.

```python
import sys
import os

# Ensure src/ is on the path
sys.path.insert(0, os.path.abspath('..'))

from src.parsing import NLUProblemParser
```

```python
# Initialize the pipeline parser
parser = NLUProblemParser(confidence_threshold=0.7)

# Exact Concrete Scenario Input Command from problem statement specifications
sample_query = "I have a telecom customer dataset with 10,000 rows. I want to predict which customers will churn next month."

print("--- Running Phase 1: NLU Intent Routing ---")
intent_results = parser.classify_intent(sample_query)
print(f"Query: {intent_results['query']}")
print(f"Routed Task Track: {intent_results['predicted_task']}")
print(f"System Confidence: {intent_results['confidence_score']} (Threshold: 0.7)")
print(f"Routing Pipeline Status: {intent_results['status']}")
print(f"Message: {intent_results['message']}")

print("\n--- Running Entity & Domain Extraction ---")
meta_results = parser.extract_metadata(sample_query)
print(f"Extracted Field Keywords: {meta_results['target_keywords']}")
print(f"Determined Domain Footprint: {meta_results['domain_context']}")
print(f"User Material Layout Profile: {meta_results['detected_expertise']}")
```
